This notebook is a clean, non-authoritative reproduction entry point. It calls the checked-in synthetic roundtrip script rather than duplicating model logic. ## Safety and authentication boundary

Never paste `HF_TOKEN`, `RUNPOD_API_KEY`, Google credentials, or other secrets into this notebook. The smoke path does not need credentials. If a future public gated-asset reproduction needs a token, provide it through the runtime's secret store (for example, Colab Secrets) and never print, save, or upload its value. Drive synchronization is a separate optional operation and is not attempted here.

```python
from pathlib import Path
import os
import subprocess
import sys

# Run from a checkout in Colab or from the repository root locally.
candidates = [Path.cwd(), Path('/content/OcuForge'), Path('/content/ocuforge')]
repo_root = next((p for p in candidates if (p / 'scripts/synthetic_roundtrip.py').is_file()), None)
if repo_root is None:
    raise FileNotFoundError('Checkout not found; clone the approved repository revision first.')

smoke_script = repo_root / 'scripts/synthetic_roundtrip.py'
protocol = repo_root / 'eyes-detected-contracts/protocols/ico_2017_dr.v0.1.json'
assert protocol.is_file(), protocol
print(f'Repository checkout found: {repo_root}')
print(f'Reproduction entry point: {smoke_script}')
print('Authentication attempted: no')
```

```python
# Keep the default smoke CPU-only and opt-in. A standby GPU is never required.
run_smoke = os.environ.get('OCUFORGE_RUN_SMOKE', '0') == '1'
if run_smoke:
    subprocess.run([sys.executable, str(smoke_script)], cwd=repo_root, check=True)
    print('Synthetic roundtrip: PASS')
else:
    print('Synthetic roundtrip: NOT_RUN (set OCUFORGE_RUN_SMOKE=1 for an explicit local/Colab run)')
```
